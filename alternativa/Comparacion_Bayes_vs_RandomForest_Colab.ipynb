# Red bayesiana vs. Random Forest
### Sistema de Recomendación Vocacional · Maestría en IA Aplicada (UDLA)

**Objetivo:** comparar el modelo elegido para el proyecto (**red bayesiana**) con la alternativa más común de
aprendizaje automático (**Random Forest**), con los mismos datos y el mismo escenario de evaluación de la tesis.

| | Red bayesiana (elegida) | Random Forest (alternativa) |
|---|---|---|
| Qué es | Modelo probabilístico: combina notas y CHASIDE según la confiabilidad de cada fuente | Conjunto de árboles de decisión que aprende reglas a partir de ejemplos |
| Para aprender necesita | Solo notas y CHASIDE (**sin respuestas correctas**) | Ejemplos con la **respuesta correcta** (etiquetas) |
| Incertidumbre | Exacta: rango y probabilidad de cada zona | Aproximada con la dispersión entre árboles |
| Explica cada resultado | Sí, por fuente | No |

**Cómo usarlo:** menú *Entorno de ejecución → Ejecutar todo* (tarda 1–2 minutos). No hay que subir ningún archivo.

```python
#@title 1 · Preparar el entorno (no editar) { display-mode: "form" }
import base64, io, os, sys, zipfile, warnings
warnings.filterwarnings('ignore')
DESTINO = '/content/comparacion' if os.path.isdir('/content') else os.path.abspath('comparacion_colab')
PAQUETE = ''
zipfile.ZipFile(io.BytesIO(base64.b64decode(PAQUETE))).extractall(DESTINO)
sys.path.insert(0, DESTINO); sys.path.insert(0, os.path.join(DESTINO, 'alternativa'))
import json, numpy as np, pandas as pd, matplotlib.pyplot as plt
from scipy.stats import norm
import comparacion as cmp
from modelo import bayes, configuracion as cfg
claves = cfg.cargar_claves(); fuentes = cfg.cargar_fuentes(); lista_car, _ = cfg.cargar_carreras()
calib_car = json.load(open(f'{DESTINO}/config/calibracion_referencia.json', encoding='utf-8'))['carreras']
AZUL, NARANJA, GRIS = '#2a78d6', '#eb6834', '#52514e'
plt.rcParams.update({'figure.dpi': 110, 'axes.spines.top': False, 'axes.spines.right': False, 'font.size': 11})
print('✓ Listo:', len(claves), 'preguntas CHASIDE ·', len(lista_car), 'carreras')
```

## 2 · El escenario de evaluación (el mismo de la tesis, Módulo 3B)
500 estudiantes con su **rendimiento real** en notas y un **CHASIDE simulado** del que conocemos la verdad
(el interés "real" de cada estudiante). Así podemos medir quién se acerca más a la verdad.
Se usa 70 % para entrenar y 30 % (150 estudiantes) para probar.

Puedes cambiar el escenario: **realista** (el oficial) o **estres** (datos más ruidosos y difíciles).

```python
escenario = "realista" #@param ["realista", "estres"]
rend, por_anio = cmp.cargar_notas()
apps, verdad, aleatorio = cmp.simular(rend, por_anio, claves, escenario)
EV, CRIT, cal = cmp.evidencia(rend, apps, claves)
train, test = map(sorted, bayes.particion(EV.keys()))
print(f'Escenario «{escenario}»: {len(EV)} estudiantes · entrenamiento {len(train)} · prueba {len(test)}')
```

## 3 · Entrenar los dos modelos
- **Red bayesiana:** aprende sus parámetros **sin mirar la respuesta correcta** (igual que en el colegio).
- **Random Forest:** 300 árboles, entrenado **con la respuesta correcta**. Es una **ventaja** que en la vida real
  no tendría: nadie conoce el "interés verdadero" de un estudiante.

```python
P = cmp.bayes_ajustar(EV, por_anio, train, cal)
pred_bayes = cmp.bayes_predecir(EV, CRIT, P, cal, test, fuentes)
pred_rf = cmp.rf_predecir(cmp.rf_ajustar(EV, CRIT, verdad, train), EV, CRIT, test)
print('✓ Red bayesiana entrenada SIN etiquetas')
print('✓ Random Forest entrenado CON etiquetas (ventaja)')
```

## 4 · Resultados: ¿cuál es mejor?

```python
METRICAS = [  # clave, nombre, qué es mejor
    ('r_I', 'Precisión del interés (correlación)', 'alto'),
    ('r_R', 'Precisión del rendimiento (correlación)', 'alto'),
    ('RMSE_I', 'Error del interés (RMSE)', 'bajo'),
    ('exactitud_zona_todas', 'Zona correcta (todas las áreas)', 'alto'),
    ('cobertura90_I', 'Rango «90 %» contiene el interés real', '90'),
    ('cobertura90_R', 'Rango «90 %» contiene el rendimiento real', '90'),
    ('brier', 'Error de las probabilidades de zona (Brier)', 'bajo'),
    ('top1_en_top3', 'Área de mayor interés dentro de su top 3', 'alto'),
    ('top5_en_comun', 'Carreras del top 5 que coinciden con el real (de 5)', 'alto'),
]
mb = {**cmp.metricas(pred_bayes, verdad), **cmp.ranking_carreras(pred_bayes, verdad, lista_car, calib_car)}
mr = {**cmp.metricas(pred_rf, verdad), **cmp.ranking_carreras(pred_rf, verdad, lista_car, calib_car)}
filas = []
for k, nombre, mejor in METRICAS:
    b, r = mb[k], mr[k]
    db, dr = (abs(b - .9), abs(r - .9)) if mejor == '90' else ((-b, -r) if mejor == 'alto' else (b, r))
    gana = 'Empate' if abs(db - dr) < 0.01 else ('Red bayesiana' if db < dr else 'Random Forest')
    filas.append({'Métrica': nombre, 'Red bayesiana': b, 'Random Forest': r, 'Mejor': gana})
tabla = pd.DataFrame(filas).set_index('Métrica')
def color(fila):
    c = {'Red bayesiana': 'background-color:#dbe9fa;font-weight:bold', 'Random Forest': 'background-color:#fde3d5;font-weight:bold'}
    return [c.get(fila['Mejor'], '') if col in (fila['Mejor'], 'Mejor') else '' for col in fila.index]
g = tabla['Mejor'].value_counts()
print(f"Red bayesiana gana {g.get('Red bayesiana', 0)} · Random Forest gana {g.get('Random Forest', 0)} · Empates {g.get('Empate', 0)}")
tabla.style.apply(color, axis=1).format({'Red bayesiana': '{:.3f}', 'Random Forest': '{:.3f}'})
```

**Cómo leerlo.** En precisión los dos son prácticamente iguales, **aunque el Random Forest vio la respuesta
correcta**. La diferencia grande está en la incertidumbre: cuando el Random Forest dice "rango del 90 %",
acierta mucho menos de lo que promete (es **sobreconfiado**). La red bayesiana cumple su 90 %.

```python
fig, ax = plt.subplots(figsize=(8, 4))
x = np.arange(2); w = .36
vb = [mb['cobertura90_I'] * 100, mb['cobertura90_R'] * 100]; vr = [mr['cobertura90_I'] * 100, mr['cobertura90_R'] * 100]
for dx, vals, col, lab in ((-w/2, vb, AZUL, 'Red bayesiana'), (w/2, vr, NARANJA, 'Random Forest')):
    barras = ax.bar(x + dx, vals, w * .95, color=col, label=lab)
    for b_, v in zip(barras, vals): ax.text(b_.get_x() + b_.get_width()/2, v + 1.5, f'{v:.0f} %', ha='center')
ax.axhline(90, color=GRIS, ls='--', lw=1); ax.text(1.55, 91.5, 'lo prometido: 90 %', ha='right', color=GRIS, fontsize=10)
ax.set_xticks(x, ['Interés', 'Rendimiento']); ax.set_ylim(0, 112); ax.set_ylabel('% de casos dentro del rango')
ax.set_title('¿El rango «90 %» contiene el valor real?', loc='left'); ax.legend(frameon=False, ncol=2, loc='upper left')
plt.tight_layout(); plt.show()
```

## 5 · ¿Y si el colegio tiene pocos estudiantes?
Entrenamos los dos modelos con 20, 50, 100, 200 y 350 estudiantes y medimos el error de las probabilidades de
zona (Brier: **menor es mejor**).

```python
rng = np.random.default_rng(1); filas = []
for n in (20, 50, 100, 200, 350):
    for rep in range(3 if n < 350 else 1):
        sub = sorted(rng.choice(train, size=n, replace=False)) if n < len(train) else train
        b = cmp.metricas(cmp.bayes_predecir(EV, CRIT, cmp.bayes_ajustar(EV, por_anio, sub, cal), cal, test, fuentes), verdad)
        r = cmp.metricas(cmp.rf_predecir(cmp.rf_ajustar(EV, CRIT, verdad, sub, semilla=rep), EV, CRIT, test), verdad)
        filas += [{'n': n, 'modelo': 'Red bayesiana', **b}, {'n': n, 'modelo': 'Random Forest', **r}]
curva = pd.DataFrame(filas).groupby(['n', 'modelo'])['brier'].mean().unstack()
fig, ax = plt.subplots(figsize=(8, 4))
for mod, col in (('Red bayesiana', AZUL), ('Random Forest', NARANJA)):
    ax.plot(curva.index, curva[mod], marker='o', lw=2, color=col, label=mod)
    ax.annotate(f'{curva[mod].iloc[0]:.2f}', (curva.index[0], curva[mod].iloc[0]), textcoords='offset points', xytext=(0, 8), ha='center', color=GRIS)
ax.set_ylim(0.2, curva.values.max() + 0.05); ax.set_xscale('log'); ax.set_xticks(curva.index, [str(n) for n in curva.index]); ax.minorticks_off()
ax.set_xlabel('Estudiantes usados para entrenar'); ax.set_ylabel('Error de Brier (menor = mejor)')
ax.set_title('Calidad de la probabilidad de cada zona según cuántos datos hay', loc='left'); ax.legend(frameon=False)
plt.tight_layout(); plt.show()
curva.round(3)
```

La red bayesiana rinde **igual con 20 estudiantes que con 350**, porque su estructura viene del conocimiento
experto. El Random Forest necesita muchos ejemplos etiquetados y aun con 350 no la alcanza.

## 6 · El mismo estudiante con los dos modelos
Elige un estudiante de prueba con el deslizador. El rombo negro es su **interés real**; las barras son el
**rango del 90 %** que da cada modelo. Un buen modelo tiene el rombo **dentro** de su barra.

```python
estudiante = 59 #@param {type:"slider", min:0, max:149, step:1}
id_est = test[estudiante]
b = pred_bayes[pred_bayes.id == id_est].set_index('area'); r = pred_rf[pred_rf.id == id_est].set_index('area')
pct = lambda z: 100 * norm.cdf(z)
NOMBRE_AREA = {'C': 'C · Administrativas', 'H': 'H · Humanísticas', 'A': 'A · Artísticas', 'S': 'S · Salud',
               'I': 'I · Ingenierías', 'D': 'D · Defensa', 'E': 'E · Ciencias exactas'}
fig, ax = plt.subplots(figsize=(9, 5)); dentro_b = dentro_r = 0
for k, a in enumerate(cmp.CODIGOS):
    real = verdad[(id_est, a)][0]; y = len(cmp.CODIGOS) - 1 - k
    lb, hb = b.loc[a, 'mu_I'] - 1.645 * b.loc[a, 'sd_I'], b.loc[a, 'mu_I'] + 1.645 * b.loc[a, 'sd_I']
    lr, hr = r.loc[a, 'lo_I'], r.loc[a, 'hi_I']
    dentro_b += lb <= real <= hb; dentro_r += lr <= real <= hr
    ax.plot([pct(lb), pct(hb)], [y + .15] * 2, color=AZUL, lw=6, solid_capstyle='round', label='Red bayesiana (rango 90 %)' if k == 0 else None)
    ax.plot([pct(lr), pct(hr)], [y - .15] * 2, color=NARANJA, lw=6, solid_capstyle='round', label='Random Forest (rango 90 %)' if k == 0 else None)
    ax.scatter(pct(real), y, marker='D', s=70, color='#0b0b0b', zorder=5, label='Interés real' if k == 0 else None)
ax.set_yticks(range(len(cmp.CODIGOS)), [NOMBRE_AREA.get(a, a) for a in cmp.CODIGOS][::-1])
ax.set_xlim(0, 100); ax.set_xlabel('Interés (percentil)')
ax.set_title(f'Estudiante {id_est}: valor real dentro del rango → bayesiana {dentro_b}/7 · Random Forest {dentro_r}/7', loc='left', fontsize=11)
ax.legend(frameon=False, loc='upper center', bbox_to_anchor=(.5, -.13), ncol=3, fontsize=9)
plt.tight_layout(); plt.show()
```

```python
#@title En todos los estudiantes de prueba { display-mode: "form" }
def cubre(df, modelo):
    d = df.copy(); d['real'] = [verdad[(i, a)][0] for i, a in zip(d.id, d.area)]
    lo, hi = (d.mu_I - 1.645 * d.sd_I, d.mu_I + 1.645 * d.sd_I) if modelo == 'b' else (d.lo_I, d.hi_I)
    return ((d.real >= lo) & (d.real <= hi)).groupby(d.id).sum()
cb, cr = cubre(pred_bayes, 'b'), cubre(pred_rf, 'r')
print(f'Estudiantes con sus 7 áreas bien cubiertas → Red bayesiana: {(cb == 7).sum()} de {len(cb)} · Random Forest: {(cr == 7).sum()} de {len(cr)}')
```

## 7 · Conclusión: por qué la red bayesiana es la mejor opción

| Criterio | Red bayesiana | Random Forest |
|---|---|---|
| Se puede usar hoy en el colegio | ✅ Sí, no necesita etiquetas | ❌ No, necesita respuestas correctas que no existen |
| Precisión | ✅ Igual o mejor | Igual (con ventaja) |
| Incertidumbre honesta (rango 90 %) | ✅ Se cumple (~88–91 %) | ❌ Sobreconfiado (~68 %) |
| Con pocos estudiantes | ✅ Estable desde 20 | ❌ Se degrada |
| Explica cada resultado | ✅ Por fuente (notas, intereses, autopercepción) | ❌ Caja negra |

**Por eso el proyecto usa la red bayesiana.** El Random Forest podría servir a futuro como *complemento*, si el
colegio acumula datos de seguimiento de egresados (carrera elegida y satisfacción).
